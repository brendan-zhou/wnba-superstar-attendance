```python
import numpy as np
import pandas as pd
import statsmodels.formula.api as smf

# Set seed for reproducibility
np.random.seed(42)

n_units = 100  # Number of units
n_periods = 10  # Number of time periods

# Create panel data (each unit appears in every time period)
df = pd.DataFrame([(i, t) for i in range(n_units) for t in range(n_periods)], columns=["unit", "time"])

# Assign staggered treatment times to each unit (some units never treated)
treatment_times = np.random.choice([3, 5, 7, np.inf], size=n_units)

# Merge treatment times back to full dataset
df = df.merge(pd.DataFrame({"unit": range(n_units), "treatment_time": treatment_times}), on="unit")

# Define treatment indicator (treated if current time is >= treatment time)
df["treated"] = (df["time"] >= df["treatment_time"]).astype(int)

# Simulate outcome variable with a treatment effect
df["outcome"] = 5 + 2 * df["treated"] + np.random.normal(0, 1, size=len(df))

# Fit a Difference-in-Differences (DiD) model
model = smf.ols("outcome ~ treated + C(unit) + C(time)", data=df).fit()
print(model.summary())
```
